
# Task 1: House Price Prediction Using Polynomial Regression

## 1. Project Objective
The goal of this project is to predict house prices based on house area using Polynomial Regression.

## 2. What Is Polynomial Regression?
Polynomial Regression is a regression technique that can learn curved relationships between input and output. It adds extra features such as Area² so the model can learn more than a straight-line relationship.

## 3. Dataset
We will create a small sample dataset inside Python, so no external file is required.

- **Input (X):** House area in square feet.
- **Target (y):** House price in thousands of dollars.

## 4. Code Explanation

1. **NumPy:** Helps store and work with numerical data.
2. **Train-test split:** Divides data into training and testing parts.
3. **PolynomialFeatures:** Creates extra features such as Area².
4. **LinearRegression:** Learns the relationship between area and price using the polynomial features.
5. **fit():** Trains the model on training data.
6. **predict():** Predicts prices for test data.
7. **MAE:** Measures the average absolute prediction error.
8. **MSE:** Measures the average squared prediction error.
9. **R² Score:** Measures how well the model explains variations in the target. A negative score is possible when predictions perform poorly.

## 5. Conclusion
We will evaluate the model using test data to understand how well it predicts house prices.


```python
%pip install scikit-learn
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Step 1: Create a sample house price dataset
area = np.array([500, 600, 700, 800, 900, 1000, 1100, 1200,
                 1300, 1400, 1500, 1600, 1700, 1800, 1900]).reshape(-1, 1)

price = np.array([80, 95, 110, 130, 155, 180, 210, 245,
                  280, 320, 365, 410, 460, 515, 575])

# Step 2: Divide data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    area, price, test_size=0.2, random_state=42
)

# Step 3: Create polynomial features of degree 2
poly = PolynomialFeatures(degree=2)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

# Step 4: Create and train the model
model = LinearRegression()
model.fit(X_train_poly, y_train)

# Step 5: Predict house prices
y_pred = model.predict(X_test_poly)

# Step 6: Evaluate the model
print("Polynomial Regression - House Price Prediction")
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Actual prices:", y_test)
print("Predicted prices:", np.round(y_pred, 2))
print("MAE:", round(mean_absolute_error(y_test, y_pred), 2))
print("MSE:", round(mean_squared_error(y_test, y_pred), 2))
print("R2 Score:", round(r2_score(y_test, y_pred), 2))
```
